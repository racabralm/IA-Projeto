```python
"""
PROJETO DE INTELIGÊNCIA ARTIFICIAL - N1

Integrantes:
- Rafael Araujo Cabral Moreira (RA: 10441919)
- Rute Willemann (RA: 10436781)

Síntese do arquivo:
Script em Python para coleta de dados financeiros originais via biblioteca Yahoo Finance.
Este código extrai, estrutura e prepara o histórico diário de ações de grandes
empresas brasileiras (Série Temporal) para a análise exploratória e futura modelagem preditiva.

Histórico de Alterações:
- 13/09/2026 - Rafael Araujo Cabral Moreira - Criação do script de captura de dados financeiros e integração com Drive.
"""

import os
import pandas as pd
from google.colab import drive

# 1. Instala a biblioteca do Yahoo Finance no Colab
os.system('pip install yfinance')
import yfinance as yf

# 2. Conecta o Colab ao seu Google Drive
try:
    drive.mount('/content/drive')
except:
    pass

print("Iniciando a captura dos dados da Bolsa de Valores...")

# 3. Define as ações que vamos usar (Petrobras, Vale, Itaú, Bradesco, Ambev)
tickers = ["PETR4.SA", "VALE3.SA", "ITUB4.SA", "BBDC4.SA", "ABEV3.SA"]
dados_acoes = []

# Captura os últimos 2 anos de dados
for ticker in tickers:
    acao = yf.Ticker(ticker)
    df_hist = acao.history(period="2y")
    df_hist['Ticker'] = ticker  # Cria uma coluna para sabermos de qual empresa é o dado
    df_hist.reset_index(inplace=True)
    dados_acoes.append(df_hist)

# 4. Junta tudo em um único DataFrame
df_final = pd.concat(dados_acoes, ignore_index=True)

# Remove o fuso horário da coluna de Data para evitar problemas no CSV
df_final['Date'] = df_final['Date'].dt.tz_localize(None)

# 5. Define o caminho exato e salva
caminho_pasta = '/content/drive/MyDrive/2026.2/IA/# Projeto'
nome_arquivo = 'dataset_acoes_negocios.csv'
caminho_final = os.path.join(caminho_pasta, nome_arquivo)

# Cria a pasta caso ela não exista e salva o CSV
os.makedirs(caminho_pasta, exist_ok=True)
df_final.to_csv(caminho_final, index=False)

print(f"Sucesso absoluto! O dataset com {len(df_final)} linhas reais foi salvo em: {caminho_final}")
```
